```python
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
from google.colab import userdata
import datasets
```

```python
tokenizer = AutoTokenizer.from_pretrained("unicamp-dl/translation-en-pt-t5")

model = AutoModelForSeq2SeqLM.from_pretrained("unicamp-dl/translation-en-pt-t5")
```

```python
def translateEN_PT(question):
  text = f"translate English to Portuguese: {question}"

  inputs = tokenizer(text, return_tensors="pt")

  outputs = model.generate(
      **inputs,
      max_new_tokens=250
  )

  result = tokenizer.decode(outputs[0], skip_special_tokens=True)
  return result
```

```python
translateEN_PT("List the names of all establishments along with the name of their city.")
```
